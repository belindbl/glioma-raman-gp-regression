## Data exploration and patient-level validation splits

Prepares a stratified five-fold cross-validation split where:

- `class_id` 1-3 -> IDHmutant
- `class_id` 4-6 -> IDHwildtype

The dataset contains sample entries, each with a 3D Raman array. A patient can have multiple entries, and each entry contains many spectra. The patient is therefore the split unit: and every entry and spectrum from one patient must stay in the same fold.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import StratifiedKFold

PATH = r"..\data\preprocessed.npy"
dataset = np.load(PATH, allow_pickle=True)

print(f"Sample entries: {len(dataset)}")
```

### Sample-level metadata table

Each top-level dataset item is treated as one sample entry. The number of spectra is the product of the first two dimensions of `data` (for example, `(121,51,1167)` contains `121 x 51` spectra, each with 1167 channels, following the pre-processing done by Lita et al.)

```python
records = pd.DataFrame([
    {
        "entry_index": i,
        "patient_id": item["patient_id"],
        "sample_name": item["name"],
        "class_id": item["class_id"],
        "n_spectra": int(np.prod(item["data"].shape[:-1])),
        "n_channels": item["data"].shape[-1],
    }
    for i, item in enumerate(dataset)
])

records.head()
```

## Binary target mapping

```python
target_by_class = {
    1: "IDHmutant",
    2: "IDHmutant",
    3: "IDHmutant",
    4: "IDHwildtype",
    5: "IDHwildtype",
    6: "IDHwildtype",
}

records["target"] = records["class_id"].map(target_by_class)
assert records["target"].notna().all(), "Unexpected or missing class_id found."

class_counts_per_patient = records.groupby("patient_id")["class_id"].nunique()
assert class_counts_per_patient.eq(1).all(), "At least one patient has multiple class_id values."

records.head()
```

## Summary at the patient level

```python
patients = (
    records.groupby("patient_id", as_index=False)
    .agg(
        target=("target", "first"),
        n_samples=("entry_index", "size"),
        n_spectra=("n_spectra", "sum"),
    )
)

cohort_summary = (
    patients.groupby("target")
    .agg(
        patients=("patient_id", "nunique"),
        sample_entries=("n_samples", "sum"),
        spectra=("n_spectra", "sum"),
    )
    .reindex(["IDHmutant", "IDHwildtype"])
)

cohort_summary = cohort_summary.reset_index() # fixed misaligned table formatting
display(cohort_summary)
print(f"Unique patients: {len(patients)}")
```

```python
print(cohort_summary.shape)
print(cohort_summary.index.tolist())
print(records["target"].value_counts(dropna=False))
```

```python
N_SPLITS = 5
RANDOM_STATE = 42

patients_per_class = patients["target"].value_counts()
assert patients_per_class.min() >= N_SPLITS, ( # Set minimum num of patients to at least the number of splits
    "Requires at least five patients in each target group."
)

splitter = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE,
)

patients["fold"] = -1
for fold, (_, validation_positions) in enumerate(
    splitter.split(patients["patient_id"], patients["target"])
):
    patients.loc[validation_positions, "fold"] = fold

assert patients["fold"].between(0, N_SPLITS - 1).all()
patients.sort_values(["fold", "target", "patient_id"]).reset_index(drop=True)
```

```python
fold_rows = []
for fold in range(N_SPLITS):
    patient_fold = patients[patients["fold"] == fold]
    entry_fold = records[records["patient_id"].isin(patient_fold["patient_id"])]
    fold_rows.append({
        "fold": fold + 1,
        "validation_patients": len(patient_fold),
        "IDHmutant_patients": int((patient_fold["target"] == "IDHmutant").sum()),
        "IDHwildtype_patients": int((patient_fold["target"] == "IDHwildtype").sum()),
        "validation_sample_entries": len(entry_fold),
        "validation_spectra": int(entry_fold["n_spectra"].sum()),
    })

fold_summary = pd.DataFrame(fold_rows).set_index("fold")
display(fold_summary)
```

## Average spectra by IDH status

Compute spectrum-weighted channel-wise means for IDH-mutant and IDH-wildtype samples. The difference spectrum is defined as mutant minus wildtype.

```python
targets = ("IDHmutant", "IDHwildtype")
channel_count = int(records["n_channels"].iloc[0])
intensity_sums = {target: np.zeros(channel_count, dtype=np.float64) for target in targets}
spectrum_counts = {target: 0 for target in targets}

# Accumulate per-entry sums to avoid concatenating all spectra into one large array.
for entry_index, item in enumerate(dataset):
    target = records.loc[records["entry_index"] == entry_index, "target"].iloc[0]
    spectra = np.asarray(item["data"]).reshape(-1, channel_count)
    intensity_sums[target] += spectra.sum(axis=0)
    spectrum_counts[target] += spectra.shape[0]

mean_spectra = {
    target: intensity_sums[target] / spectrum_counts[target]
    for target in targets
}
difference_spectrum = mean_spectra["IDHmutant"] - mean_spectra["IDHwildtype"]
mean_signed_difference = float(difference_spectrum.mean())
mean_absolute_difference = float(np.abs(difference_spectrum).mean())

print(f"Spectra averaged: {spectrum_counts}")
print(f"Mean signed difference (mutant - wildtype): {mean_signed_difference:.6g}")
print(f"Mean absolute difference across channels: {mean_absolute_difference:.6g}")

channel_index = np.arange(channel_count)
fig, (ax_means, ax_difference) = plt.subplots(2, 1, figsize=(14, 9), sharex=True)
ax_means.plot(channel_index, mean_spectra["IDHmutant"], label="IDH-mutant")
ax_means.plot(channel_index, mean_spectra["IDHwildtype"], label="IDH-wildtype")
ax_means.set_ylabel("Mean intensity")
ax_means.set_title("Average Raman spectra by IDH status")
ax_means.legend()
ax_means.grid(alpha=0.25)

ax_difference.plot(channel_index, difference_spectrum, color="black")
ax_difference.axhline(0, color="gray", linewidth=1, linestyle="--")
ax_difference.set_xlabel("Channel index")
ax_difference.set_ylabel("Mean intensity difference")
ax_difference.set_title("Difference spectrum (IDH-mutant - IDH-wildtype)")
ax_difference.grid(alpha=0.25)
fig.tight_layout()
plt.show()
```

## PCA of spectra by IDH status and patient

Fit PCA using only spectral channel intensities. Patient IDs are used to balance the sampled spectra and calculate patient mean centroids; IDH status is used only to color and compare those spectral scores. Neither metadata field is included as a PCA feature.


```python
from sklearn.decomposition import PCA

PCA_SPECTRA_PER_PATIENT = 5000
pca_rng = np.random.default_rng(42)
channel_count = int(records['n_channels'].iloc[0])
patient_groups = records.groupby('patient_id', sort=True)
sampled_spectra, sampled_patient_ids, sampled_statuses = [], [], []
patient_mean_sums, patient_spectrum_counts = {}, {}

for patient_id, entries in patient_groups:
    status = entries['target'].iloc[0]
    entry_indices = entries['entry_index'].to_numpy(dtype=int)
    patient_mean_sums[patient_id] = np.zeros(channel_count, dtype=np.float64)
    patient_spectrum_counts[patient_id] = 0
    base, extra = divmod(PCA_SPECTRA_PER_PATIENT, len(entry_indices))
    
    for position, entry_index in enumerate(entry_indices):
        spectra = np.asarray(dataset[entry_index]['data']).reshape(-1, channel_count)
        patient_mean_sums[patient_id] += spectra.sum(axis=0)
        patient_spectrum_counts[patient_id] += len(spectra)
        quota = base + (position < extra)
        
        if quota:
            chosen = pca_rng.choice(len(spectra), size=min(quota, len(spectra)), replace=False)
            sampled_spectra.append(spectra[chosen])
            sampled_patient_ids.extend([patient_id] * len(chosen))
            sampled_statuses.extend([status] * len(chosen))

# PCA is fit to spectral channels only; IDs and status remain annotations.
pca_input = np.vstack(sampled_spectra)
sampled_patient_ids = np.asarray(sampled_patient_ids)
sampled_statuses = np.asarray(sampled_statuses)
patient_ids = np.asarray(list(patient_groups.groups.keys()))
patient_statuses = np.asarray([patient_groups.get_group(pid)['target'].iloc[0] for pid in patient_ids])
patient_means = np.vstack([patient_mean_sums[pid] / patient_spectrum_counts[pid] for pid in patient_ids])
pca = PCA(n_components=10, svd_solver='randomized', random_state=42)
sample_scores = pca.fit_transform(pca_input)
patient_centroid_scores = pca.transform(patient_means)

print(f'PCA sample: {len(pca_input):,} spectra ({PCA_SPECTRA_PER_PATIENT} per patient when available)')
print('Explained variance ratio, first 5 PCs:', np.round(pca.explained_variance_ratio_[:5], 4))

colors = {'IDHmutant': 'tab:blue', 'IDHwildtype': 'tab:orange'}
component_pairs = [(0, 1), (2, 3), (4, 5), (6, 7)]
fig, axes = plt.subplots(nrows=4, ncols=1, figsize=(12, 28))

for ax, (first, second) in zip(axes, component_pairs):
    for status, color in colors.items():
        mask = sampled_statuses == status
        ax.scatter(sample_scores[mask, first], sample_scores[mask, second], s=12, alpha=0.18, color=color, label=f'{status} spectra', rasterized=True)

    for status, color in colors.items():
        mask = patient_statuses == status
        ax.scatter(patient_centroid_scores[mask, first], patient_centroid_scores[mask, second], s=85, marker='D', edgecolor='black', linewidth=0.8, color=color, label=f'{status} patient mean spectrum', zorder=3)
        for pid, x, y in zip(patient_ids[mask], patient_centroid_scores[mask, first], patient_centroid_scores[mask, second]):
            ax.annotate(str(pid), (x, y), xytext=(4, 4), textcoords='offset points', fontsize=8)

    ax.set_xlabel(f'PC{first + 1} ({pca.explained_variance_ratio_[first]:.1%} variance)')
    ax.set_ylabel(f'PC{second + 1} ({pca.explained_variance_ratio_[second]:.1%} variance)')
    ax.set_title(f'PC{first + 1} vs PC{second + 1}')
    ax.grid(alpha=0.2)
    ax.legend()

fig.suptitle('PCA of spectral channel intensities', y=1.002)
fig.tight_layout()
plt.show()
```

### Follow-up on PC8

Compare patient mean-spectrum scores on PC8 between IDH groups, treating each patient as one observation, and inspect the PC8 loadings across spectral channels. The loading sign is arbitrary: PCA can flip the signs of both scores and loadings without changing the component. Channel indices are shown because the dataset does not provide a wavelength axis here.


```python
from scipy.stats import ttest_ind

pc8_index = 7
pc8_scores_by_status = {
    status: patient_centroid_scores[patient_statuses == status, pc8_index]
    for status in ('IDHmutant', 'IDHwildtype')
}
mutant_pc8 = pc8_scores_by_status['IDHmutant']
wildtype_pc8 = pc8_scores_by_status['IDHwildtype']
welch_result = ttest_ind(wildtype_pc8, mutant_pc8, equal_var=False)
mean_difference = wildtype_pc8.mean() - mutant_pc8.mean()
pooled_sd = np.sqrt((mutant_pc8.var(ddof=1) + wildtype_pc8.var(ddof=1)) / 2)
cohens_d = mean_difference / pooled_sd if pooled_sd else np.nan

print(f"PC8 explains {pca.explained_variance_ratio_[pc8_index]:.2%} of sampled spectral variance.")
print(f"IDHmutant: n={len(mutant_pc8)}, mean={mutant_pc8.mean():.4g}, SD={mutant_pc8.std(ddof=1):.4g}")
print(f"IDHwildtype: n={len(wildtype_pc8)}, mean={wildtype_pc8.mean():.4g}, SD={wildtype_pc8.std(ddof=1):.4g}")
print(f"Mean difference (wildtype - mutant)={mean_difference:.4g}; Welch t={welch_result.statistic:.3f}, p={welch_result.pvalue:.4g}; Cohen's d={cohens_d:.3f}")

fig, (ax_scores, ax_loadings) = plt.subplots(2, 1, figsize=(14, 10))
status_order = ['IDHmutant', 'IDHwildtype']
score_data = [pc8_scores_by_status[status] for status in status_order]
ax_scores.boxplot(score_data, showfliers=False)
for x, status in enumerate(status_order, start=1):
    values = pc8_scores_by_status[status]
    jitter = np.random.default_rng(42 + x).uniform(-0.08, 0.08, size=len(values))
    ax_scores.scatter(x + jitter, values, alpha=0.8, s=35, label=status)
ax_scores.axhline(0, color='gray', linewidth=1, linestyle='--')
ax_scores.set_ylabel('Patient mean spectrum PC8 score')
ax_scores.set_title('PC8 scores by IDH status (one point per patient)')
ax_scores.grid(axis='y', alpha=0.2)

pc8_loadings = pca.components_[pc8_index]
channel_index = np.arange(len(pc8_loadings))
ax_loadings.plot(channel_index, pc8_loadings, color='black')
ax_loadings.axhline(0, color='gray', linewidth=1, linestyle='--')
ax_loadings.set_xlabel('Spectral channel index')
ax_loadings.set_ylabel('PC8 loading')
ax_loadings.set_title('PC8 loadings across spectral channels')
ax_loadings.grid(alpha=0.2)
fig.tight_layout()
plt.show()

largest_positive = np.argsort(pc8_loadings)[-10:][::-1]
largest_negative = np.argsort(pc8_loadings)[:10]
print('Largest positive PC8 loadings (channel, loading):', [(int(i), float(pc8_loadings[i])) for i in largest_positive])
print('Largest negative PC8 loadings (channel, loading):', [(int(i), float(pc8_loadings[i])) for i in largest_negative])
```

Patient 1887 was labeled as an outlier in a prior study using non-GP models. The next cell checks that label against spectral data only: it ranks each patient mean spectrum by its Mahalanobis distance from the mean spectrum of patients with the same IDH status, using PC1 and PC2. The status is used only to define the comparison group; this is a descriptive check in the displayed PCA plane.


```python
from numpy.linalg import pinv
centroid_scores_2d = patient_centroid_scores[:, :2]
distance_rows = []

for status in np.unique(patient_statuses):
    mask = patient_statuses == status
    group_scores = centroid_scores_2d[mask]
    centered = group_scores - group_scores.mean(axis=0)
    inverse_covariance = pinv(np.cov(group_scores, rowvar=False))
    distances = np.sqrt(np.einsum('ij,jk,ik->i', centered, inverse_covariance, centered))
    distance_rows.extend(zip(patient_ids[mask], patient_statuses[mask], distances))

patient_distance_ranking = pd.DataFrame(distance_rows, columns=['patient_id', 'target', 'mahalanobis_distance_pc1_pc2'])
patient_distance_ranking['rank_within_status'] = patient_distance_ranking.groupby('target')['mahalanobis_distance_pc1_pc2'].rank(method='min', ascending=False).astype(int)
display(patient_distance_ranking.sort_values(['target', 'rank_within_status']).groupby('target').head(5))
patient_1887 = patient_distance_ranking.loc[patient_distance_ranking['patient_id'] == 1887]
display(patient_1887)

fig, ax = plt.subplots(figsize=(10, 8))
for status, color in colors.items():
    mask = patient_statuses == status
    ax.scatter(centroid_scores_2d[mask, 0], centroid_scores_2d[mask, 1], color=color, alpha=0.75, label=status)
    for pid, x, y in zip(patient_ids[mask], centroid_scores_2d[mask, 0], centroid_scores_2d[mask, 1]):
        if pid != 1887:
            ax.annotate(str(pid), (x, y), xytext=(3, 3), textcoords='offset points', fontsize=8, alpha=0.75)

if not patient_1887.empty:
    row = int(np.flatnonzero(patient_ids == 1887)[0])
    x, y = centroid_scores_2d[row]
    ax.scatter([x], [y], color='crimson', label='Patient 1887', zorder=5)
    ax.annotate('1887 (IDHmutant)', (x, y), xytext=(8, 8), textcoords='offset points', color='crimson', fontweight='bold')
    print(f"Patient 1887 is rank {int(patient_1887['rank_within_status'].iloc[0])} of {int((patient_statuses == patient_1887['target'].iloc[0]).sum())} {patient_1887['target'].iloc[0]} patients by PC1/PC2 spectral centroid distance.")
else:
    print('Patient 1887 was not found in the PCA patient centroids.')
    
ax.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%} variance)')
ax.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%} variance)')
ax.set_title('Patient mean spectra in PCA space')
ax.grid(alpha=0.2)
ax.legend()
fig.tight_layout()
plt.show()
```

## Average spectra and differences by validation fold

For each fold, calculate a mean spectrum per patient and then average those patient means within IDH status. This gives every validation patient equal weight. The difference spectrum is IDH-mutant minus IDH-wildtype.

```python
fold_targets = ('IDHmutant', 'IDHwildtype')
channel_count = int(records['n_channels'].iloc[0])
patient_mean_spectra = {}
for patient_id, entries in records.groupby('patient_id', sort=False):
    intensity_sum = np.zeros(channel_count, dtype=np.float64)
    n_spectra = 0
    for entry_index in entries['entry_index'].to_numpy(dtype=int):
        spectra = np.asarray(dataset[entry_index]['data']).reshape(-1, channel_count)
        intensity_sum += spectra.sum(axis=0)
        n_spectra += len(spectra)
    patient_mean_spectra[patient_id] = intensity_sum / n_spectra

fold_average_spectra = {}
fold_difference_spectra = {}
fold_difference_summary = []
for fold in range(N_SPLITS):
    fold_patients = patients.loc[patients['fold'] == fold]
    group_means = {}
    for target in fold_targets:
        patient_ids_in_group = fold_patients.loc[fold_patients['target'] == target, 'patient_id'].tolist()
        group_means[target] = np.mean([patient_mean_spectra[pid] for pid in patient_ids_in_group], axis=0)
    fold_average_spectra[fold] = group_means
    difference = group_means['IDHmutant'] - group_means['IDHwildtype']
    fold_difference_spectra[fold] = difference
    fold_difference_summary.append({
        'fold': fold + 1,
        'IDHmutant_patients': int((fold_patients['target'] == 'IDHmutant').sum()),
        'IDHwildtype_patients': int((fold_patients['target'] == 'IDHwildtype').sum()),
        'mean_signed_difference': float(difference.mean()),
        'mean_absolute_difference': float(np.abs(difference).mean()),
    })

fold_difference_summary = pd.DataFrame(fold_difference_summary).set_index('fold')
display(fold_difference_summary)

channel_index = np.arange(channel_count)
fig, axes = plt.subplots(N_SPLITS, 2, figsize=(15, 20), sharex=True)
for fold, (ax_means, ax_difference) in enumerate(axes):
    means = fold_average_spectra[fold]
    ax_means.plot(channel_index, means['IDHmutant'], label='IDH-mutant', color='tab:blue')
    ax_means.plot(channel_index, means['IDHwildtype'], label='IDH-wildtype', color='tab:orange')
    ax_means.set_title(f'Fold {fold + 1}: mean spectra')
    ax_means.set_ylabel('Mean intensity')
    ax_means.grid(alpha=0.2)
    ax_means.legend()

    ax_difference.plot(channel_index, fold_difference_spectra[fold], color='black')
    ax_difference.axhline(0, color='gray', linewidth=1, linestyle='--')
    ax_difference.set_title(f'Fold {fold + 1}: mutant - wildtype')
    ax_difference.set_ylabel('Intensity difference')
    ax_difference.grid(alpha=0.2)

axes[-1, 0].set_xlabel('Channel index')
axes[-1, 1].set_xlabel('Channel index')
fig.suptitle('IDH average Raman spectra by validation fold', y=1.002)
fig.tight_layout()
plt.show()
```
